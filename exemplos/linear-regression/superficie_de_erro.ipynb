### Importando as bibliotecas.

````python
# Import all necessary libraries.
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm
````

````python
# Resetando gerador de sequências pseudo-aleatórias de forma que os resultados apresentados sejam sempre os mesmos.
np.random.seed(1234)
````

### Gerando a função observável.

A **função objetivo** é dada por

$$y = x_1 + x_2,$$

onde $a_1$ e $a_2$ são iguais a $1$ e os atributos, $x_1$ e $x_2$, são valores retirados de distribuições uniformemente distribuídas no intervalo $[-1,1)$.

**OBS**.: Percebam que a função objetivo é um **plano** (i.e., superfície em 2D).

A **função observável** (versão ruidosa da função objetivo, $y$) é dada por

$$y_{noisy} = y + w,$$

onde $w$ é o ruído Gaussiano normal padrão adicionado às amostras da função objetivo.

Nesse exemplo, geramos 1000 amostras.

````python
# Número de exemplos.
N = 1000

# Atributos variando no intervalo de -1 a 1.
x1 = 2*np.random.rand(N, 1) - 1
x2 = 2*np.random.rand(N, 1) - 1

# Ruído Gaussiano normal padrão.
w = np.random.randn(N, 1)

# Função objetivo (equação de um plano).
y = x1 + x2

# Função observável (versão ruidosa de y)
y_noisy = y + w
print(np.any(x1 == 1))

````

````python
# Plot histograms of x1 and x2 on the same axes
plt.hist(x1, bins=50, edgecolor='black', density=True, alpha=0.6, label='x1', color='red')
plt.hist(x2, bins=50, edgecolor='black', density=True, alpha=0.6, label='x2', color='blue')

# Add labels and title
plt.title("Uniform Distribution on [-1, 1) for x1 and x2")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()

# Show plot
plt.show()
````

### Comparando a função objetivo com sua versão ruidosa.

**OBS**.: Se tivéssemos mais de dois atributos, já não seria possível plotar uma figura.

````python
# Plot observable function.
fig = plt.figure(figsize=(5,5))
plt.subplot(projection='3d')
ax = fig.gca()
ax.scatter(x1, x2, y, label='Função objetivo')
ax.scatter(x1, x2, y_noisy, label='Função observável (versão ruidosa)')
ax.set_xlabel('$x_1$', fontsize=14)
ax.set_ylabel('$x_2$', fontsize=14)
ax.set_zlabel('$y$', fontsize=14)
ax.zaxis.labelpad=-3.5
plt.legend()
#plt.savefig("observable_function2.png", dpi=600)
plt.show()
````

### Aproximando a função objetivo.

E se nós quiséssemos aproximar a **função objetivo** a partir, apenas, de suas amostras ruidosas com a seguinte função hipótese:

$$h(\textbf{x}(n),\hat{\textbf{a}}) = \hat{y}(n) = \hat{a}_1 x_1(n) + \hat{a}_2 x_2(n).$$

**OBS**.: para não confundirmos com os pesos originais, $a_1$ e $a_2$, vamos usar $\hat{a}_1$ e $\hat{a}_2$ para nos referir aos pesos da função hipótese.

<span style="color:blue">Como encontraríamos os valores de $\hat{a}_1$ e $\hat{a}_2$?</span>

Até o momento conseguimos encontrá-los de duas formas:

+ Através da **equação normal**, que nos dá os **valores ótimos dos pesos**.

+ Ou **visualmente**, plotando a superfície criada a partir da **função do erro quadrático médio** (EQM):

$$J_e(\hat{a}_1, \hat{a}_2) = \frac{1}{N}\sum_{n=0}^{N-1}{(y_{noisy}(n) - \hat{y}(n))^2} \\
=\frac{1}{N}\sum_{n=0}^{N-1}{(y_{noisy}(n) - (\hat{a}_1 x_1(n) + \hat{a}_2 x_2(n)))^2}.
$$

Os valores de erro, $J_e(\hat{a}_1, \hat{a}_2)$, para plotarmos a superfície são obtidos variando-se $\hat{a}_1$ e $\hat{a}_2$ na equação do EQM.

### Encontrando a solução ótima com a equação normal.

Nós vamos encontrar a solução ótima (i.e., os pesos ótimos) e plotá-la na superfície de erro e vermos que ela realmente nos dá o ponto de menor erro possível.

A função hipótese utilizada neste exemplo é dada por

$$h(\textbf{x}(n),\hat{\textbf{a}}) = \hat{y}(n) = \hat{a}_1 x_1(n) + \hat{a}_2 x_2(n),$$

a qual pode ser re-escrita em **formato matricial** como

$$ \hat{\textbf{y}} = \textbf{X}\hat{\textbf{a}},$$

onde $\textbf{X}$ é a matriz de atributos com dimensão, neste exemplo, $N \times K$ e $\hat{\textbf{a}}$ é o vetor de pesos, com dimensão $K \times 1$.

**OBS**.: percebam que não temos o **peso de bias**, $\hat{a}_0$, pois a função objetivo não o tem também.

````python
# Concatenando os 2 vetores de atributos para criar a matriz de atributos com dimensão N x K.
X = np.c_[x1, x2]

# Equação normal.
a_opt = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y_noisy)

# Imprimindo os valores encontrados para os pesos, a1 e a2.
print('peso a1:', a_opt[0, 0])
print('peso a2:', a_opt[1, 0])

# Realizando a predição com a função hipótese em sua forma matricial.
yhat = X.dot(a_opt)

# Calculando o erro quadrático médio.
Joptimum = (1.0/N)*np.sum(np.power((y_noisy - yhat), 2))

# Imprimindo o valor do erro.
print('\nMSE:', Joptimum)
````

### Definição da função que calcula os valores para plotarmos a superfície de erro.

+ O erro é calculado variando-se $\hat{a}_1$ e $\hat{a}_2$ na equação do erro quadrático médio (EQM).
+ Essa variação é feita com as funções `linspace` e `meshgrid` da biblioteca **numpy**.
+ A função `linspace` cria uma sequência de valores linearmente espaçados dentro do intervalo definido.
+ A função `meshgrid` cria uma **malha** de valores que combina os valores de $\hat{a}_1$ e $\hat{a}_2$:

```python
a1 = np.linspace(0, 4, 5)
a2 = np.linspace(0, 4, 5)

a1 = [0, 1, 2, 3, 4]

a2 = [0, 1, 2, 3, 4]

A1, A2 = np.meshgrid(a1, a2)

A1 =  0 1 2 3 4        A2 =  0 0 0 0 0
      0 1 2 3 4              1 1 1 1 1
      0 1 2 3 4              2 2 2 2 2
      0 1 2 3 4              3 3 3 3 3
      0 1 2 3 4              4 4 4 4 4
        
plt.plot(A1, A2, marker='o', color='k')
```

<img src="https://github.com/zz4fap/t319_aprendizado_de_maquina/blob/main/figures/meshgrid.png?raw=1" width="400px">

````python
def calculateErrorSurface(x1, x2, y):
    """Generate data points for plotting the error surface."""

    # Retrieve number of examples.
    N = len(y)

    # Generate values for parameter space.
    M = 200
    a1 = np.linspace(-2.0, 4.0, M)
    a2 = np.linspace(-2.0, 4.0, M)

    # Generate matrices with combinations between a1 and a2 values.
    A1, A2 = np.meshgrid(a1, a2)

    # Generate points for plotting the cost-function surface.
    J = np.zeros((M,M))
    for iter1 in range(0, M):
        for iter2 in range(0, M):
            # Hypothesis function.
            yhat = A1[iter1, iter2]*x1 + A2[iter1, iter2]*x2
            # Calculate the mean squared error (MSE) for each pair of values.
            J[iter1, iter2] = (1.0/N)*np.sum(np.square(y - yhat))

    return J, A1, A2
````

### Plotando a superfície de erro.

+ A superfície de erro é representada por uma **figura em três dimensões**, onde cada par de valores $\hat{a}_1$ e $\hat{a}_2$ corresponde a um erro, $J_e(a_1, a_2)$.
+ Observem que a superfície de erro é **convexa** e, consequentemente, temos apenas um ponto de mínimo, chamado de **mínimo global**.

**OBS**.: O ponto ótimo, que nós encontramos com a equação normal, também é plotado na figura abaixo com uma estrela vermelha.

````python
# Invocando a função que calcula os pontos para plotarmos a superfície de erro.
J, A1, A2 = calculateErrorSurface(x1, x2, y_noisy)
# Plot cost-function surface.
fig = plt.figure(figsize=(5,5))
ax = fig.add_subplot(projection='3d')  # forma mais atual de criar eixo 3D
surf = ax.plot_surface(A1, A2, J, cmap=cm.coolwarm, linewidth=0, antialiased=False)

# Ponto ótimo
ax.plot([a_opt[0, 0]], [a_opt[1, 0]], [Joptimum],
        c='r', marker='*', markersize=10, label='Ponto ótimo')

# Labels corrigidos (usando raw string r'...')
ax.set_xlabel(r'$\hat{a}_1$')
ax.set_ylabel(r'$\hat{a}_2$')
ax.set_zlabel(r'$J_e$')

# Ajustes visuais
ax.zaxis.labelpad = -7
plt.title('Superfície da função de erro')
ax.view_init(5, 35)
plt.legend()

# Salvar em alta resolução, se desejar
# plt.savefig("error_surface_example2.png", dpi=600)

# Mostrar o plot
plt.show()
````

### Plotando a superfície de contorno.

+ Outra figura importante que plotamos a partir dos resultados obtidos para plotarmos a superfície de erro é chamada de **superfície de contorno**.
</br>

+ Ela equivale a olhar a superfície de erro por cima e **traçar planos** (paralelos aos eixos dos pesos) para diferentes valores de erro.

<img src="https://github.com/zz4fap/t319_aprendizado_de_maquina/blob/main/figures/superficie_de_contorno.png?raw=1" width="200px">

+ Cada círculo na figura é chamado de **linha de contorno**.
</br>

+ Nesse exemplo, as **linhas de contorno** são circulares, pois a superfície de erro é circular (i.e., tem formato de tigela).
</br>

+ Cada **linha de contorno** representa uma determinada elevação (i.e., erro) na superfície.
</br>

+ O **erro é constante** ao longo de cada linha de contorno.

  + Ou seja, qualquer par de valores $\hat{a}_1$ e $\hat{a}_2$ ao longo de uma mesma curva terá o mesmo valor de erro.</br>
</br>

+ Vejam que o **raio dos círculos diminui** conforme nos aproximamos do ponto de mínimo (estrela vermelha), isto indica que o erro esta diminuindo.

````python
# Figura e eixos
fig = plt.figure(figsize=(5,5))
ax = fig.add_subplot(111)

# Contorno da superfície de custo
cp = ax.contour(A1, A2, J)  # opcional: cmap=cm.coolwarm

# Ponto ótimo encontrado
ax.plot(a_opt[0], a_opt[1], marker='*', markersize=10, color='r', label='Ponto ótimo')

# Rótulos (raw strings para evitar SyntaxWarning com \hat)
ax.set_xlabel(r'$\hat{a}_1$', fontsize=14)
ax.set_ylabel(r'$\hat{a}_2$', fontsize=14)

# Rotular as linhas de contorno
ax.clabel(cp, inline=True, fontsize=10)

# Título e eixos
ax.set_title('Superfície de contorno')
ax.set_xticks(np.arange(-2, 4, step=1.0))
ax.set_yticks(np.arange(-2, 4, step=1.0))

ax.legend()
plt.tight_layout()

# Mostrar
plt.show()
````

### Plotando as funções observável e objetivo juntamente com o hiperplano encontrado com a equação normal.

````python
# Plot cost-function surface.
fig = plt.figure(figsize=(5,5))
plt.subplot(projection='3d')
ax = fig.gca()

M = 10
x1_ = np.linspace(-1.0, 1.0, M)
x2_ = np.linspace(-1.0, 1.0, M)
X1, X2 = np.meshgrid(x1_, x2_)
Y = a_opt[0]*X1 + a_opt[1]*X2

ax.scatter(x1, x2, y, label='Função objetivo')
ax.scatter(x1, x2, y_noisy, label='Função observável')
ax.plot_wireframe(X1, X2, Y, color="black", label='Função hipótese')
ax.set_xlabel('$x_1$')
ax.set_ylabel('$x_2$')
ax.set_zlabel('$y$')
plt.legend()
plt.show()
````
